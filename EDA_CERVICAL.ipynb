```python
# Binary flag says 0 (No), but duration/count is > 0
contradictions = {
    "Smokes=0 but years>0": df[(df["Smokes"] == 0) & (df["Smokes (years)"] > 0)],
    "Smokes=0 but packs>0": df[(df["Smokes"] == 0) & (df["Smokes (packs/year)"] > 0)],
    "Smokes=1 but years==0": df[(df["Smokes"] == 1) & (df["Smokes (years)"] == 0)],
    "Contraceptives=0 but years>0": df[(df["Hormonal Contraceptives"] == 0) & (df["Hormonal Contraceptives (years)"] > 0)],
    "IUD=0 but years>0": df[(df["IUD"] == 0) & (df["IUD (years)"] > 0)],
    "STDs=0 but number>0": df[(df["STDs"] == 0) & (df["STDs (number)"] > 0)],
}

for rule, bad_rows in contradictions.items():
    print(f"{rule}: {len(bad_rows)} rows")
    if len(bad_rows) > 0:
        display(bad_rows[["Age", "Smokes", "Smokes (years)", "Hormonal Contraceptives", "Hormonal Contraceptives (years)", "IUD", "IUD (years)", "STDs", "STDs (number)"]])
```

```python
numerical_columns = [
    'Age', 'Number of sexual partners', 'First sexual intercourse', 'Num of pregnancies',
    'Smokes (years)', 'Smokes (packs/year)', 'Hormonal Contraceptives (years)',
    'IUD (years)', 'STDs (number)', 'STDs: Number of diagnosis'
]
```

```python
import math

n_cols = 3
n_rows = math.ceil(len(numerical_columns) / n_cols)

fig, axes = plt.subplots(
    n_rows, n_cols,
    figsize=(16, n_rows * 4)
)

axes = axes.flatten()

for i, col in enumerate(numerical_columns):
    sns.histplot(
        df[col],
        kde=True,
        bins=30,
        ax=axes[i]
    )

    axes[i].set_title(f'Distribution of {col}')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Frequency')

# Remove unused plots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

```python
TARGET = "Biopsy"

# Diagnostic and screening results: leakage, not intake data
leakage = ["Dx", "Dx:Cancer", "Dx:CIN", "Dx:HPV",
           "Hinselmann", "Schiller", "Citology"]

# Near-duplicates
redundant = ["STDs: Number of diagnosis",   # ~ STDs (0.91)
             "STDs:vulvo-perineal condylomatosis",
             "Smokes (packs/year)",          # ~ Smokes (years) (0.72)
             "STDs (number)",
             "IUD (years)"]                  # ~ IUD (0.75)


already_dropped = ["STDs:cervical condylomatosis", "STDs:AIDS",
                   "STDs: Time since first diagnosis",
                   "STDs: Time since last diagnosis"]

drop_cols = [c for c in leakage + redundant + already_dropped if c in df.columns]
X = df.drop(columns=drop_cols + [TARGET])
y = df[TARGET]

print("Features:", X.shape[1])
print(list(X.columns))
```

```python
corr_df = X.copy()
corr_df[TARGET] = y

corr = corr_df.corr()

plt.figure(figsize=(20, 18))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm",
            cbar=True, annot_kws={"size": 7}, square=True)
plt.title("Correlation Matrix (Biopsy)")
plt.tight_layout()
plt.show()
```
